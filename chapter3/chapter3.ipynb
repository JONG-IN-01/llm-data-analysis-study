## 4. CSV 간 키 관계 검증
- 없는 `customer_id`: orders.customer_id -> customers.customer_id
- 없는 `order_id`: order_items.order_id -> orders.order_id
- 없는 `product_id`: order_items.product.id -> products.product_id

![PK FK 관계 검증](images/step04_relationship.png)

### 결과 관찰
orders.customer_id -> customers.customer_id = 0
order_items.order_id -> orders.order_id = 0
order_items.product.id -> products.product_id = 0
로 N:1 관계로 연결이 된다.

### 나의 해석과 판단
키 관계 문제가 발견되었다면 바로 삭제시 연결된 데이터들이 연쇄 삭제될 수 있기 때문에 바로 삭제하는 것은 삼가해야한다.

### 업무·분석적 의미
각 데이터 테이블들의 관계를 파악함으로써 분석하고자 하는 데이터의 내용을 효과적이며 효율적으로 파악할 수 있다.

### 한계와 추가 확인 사항
데이터 테이블을 보고 PK FK 관계를 파악하는데 시간이 소요될 수 있다.
