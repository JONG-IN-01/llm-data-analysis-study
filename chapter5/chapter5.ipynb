### 결과 관찰
실패건수는 0으로 나왔으며, 나이는 중앙값, 도시의 결측치는 unknown으로 수정하였다.

### 나의 해석과 판단
나이는 중앙값이 통계적으로 안정적이라고 판단했고, 도시의 결측치는 unknown으로 처리하는 것이 통계처리할 때 안정적이라고 생각했다. 삭제나 대체는 정보 손실을 만들 수 있다. 나이 같은 경우도 만약 실제와 다른 금액으로 대체될 가능성이 있는데, 이는 정보의 손실을 가져올 것이다. 하지만, 그것을 최소화하기 위해 중앙값이나 최빈값을 사용하는 것이다.

### 업무·분석적 의미
실제 데이터는 정리가 제대로 되어 있지 않기에 전처리 과정은 중요하다. 그렇기에 문자열을 정리하고, 숫자 변환 ,날짜 변환 같은 경우는 중요할 것이고, 실제 데이터 분석에서 통계적으로 유의미하게 영향을 미치지 않기 위해 결측치를 어떤 방법으로 처리해야할지를 잘 생각해보면서 의사결정 해야할 것이다.

### 한계와 추가 확인 사항
위에서 언급했다시피, 결측치를 대체, 삭제하는 것으로 통계적으로 영향을 미치기 때문에 영향을 최소화할 수 있는 방법을 적용해야할 것이다.

```python
def strip_string_columns(df):
    result = df.copy()
    string_columns = result.select_dtypes(include='object').columns

    for column in string_columns:
        result[column] = result[column].where(
            result[column].isna(),
            result[column].astype(str).str.strip(),
        )
        result[column] = result[column].replace('', pd.NA)

    return result

customers_clean = strip_string_columns(customers_clean)
products_clean = strip_string_columns(products_clean)
orders_clean = strip_string_columns(orders_clean)
order_items_clean = strip_string_columns(order_items_clean)

print('문자열 앞뒤 공백 제거 완료')
```

STEP 12. 전처리 후 파일 관계 재검증

```python
def validate_relationships(data: dict[str, pd.DataFrame]) -> pd.DataFrame:
    """전처리 후 파일 간 키 관계가 유지되는지 확인합니다."""
    customers = data["customers"]
    products = data["products"]
    orders = data["orders"]
    order_items = data["order_items"]

    checks = []

    if {"customer_id"}.issubset(customers.columns) and {"customer_id"}.issubset(orders.columns):
        invalid_customers = orders[~orders["customer_id"].isin(customers["customer_id"])]
        checks.append(
            {
                "check": "orders.customer_id exists in customers.customer_id",
                "invalid_count": len(invalid_customers),
            }
        )

    if {"order_id"}.issubset(orders.columns) and {"order_id"}.issubset(order_items.columns):
        invalid_orders = order_items[~order_items["order_id"].isin(orders["order_id"])]
        checks.append(
            {
                "check": "order_items.order_id exists in orders.order_id",
                "invalid_count": len(invalid_orders),
            }
        )

    if {"product_id"}.issubset(products.columns) and {"product_id"}.issubset(order_items.columns):
        invalid_products = order_items[
            ~order_items["product_id"].isin(products["product_id"])
        ]
        checks.append(
            {
                "check": "order_items.product_id exists in products.product_id",
                "invalid_count": len(invalid_products),
            }
        )

    return pd.DataFrame(checks)


processed_data = {
    "customers": customers_clean,
    "products": products_clean,
    "orders": orders_clean,
    "order_items": order_items_clean,
}

relationship_checks = validate_relationships(processed_data)
relationship_checks
```

<br>• 타 변수와의 상관관계가 반영되지 않음 |
| **B. 세그먼트별 대표값 대체**<br>

<br>*(Group-based Imputation)* | `city`, 가입 채널, 구매 등급 등 타 변수로 그룹화 후 해당 그룹의 중앙값 대체 | • 고객 특성(지역, 행동 등)을 반영하여 A안보다 정교함 | • 그룹별 샘플 수가 적을 경우 편향 발생 가능<br>

<br>• 기준 변수 선정 기준의 타당성 필요 |
| **C. 예측 모델 기반 대체**<br>

<br>*(KNN / Regression)* | 타 수치형/범주형 변수를 조건으로 머신러닝 알고리즘을 이용해 예측값 부여 | • 다변량 관계를 종합적으로 반영하여 정교한 추정 가능 | • 3건 처리에 비해 오버헤드(복잡도)가 큼<br>

<br>• 모델 결과 해석 및 유지보수가 까다로움 |
| **D. 결측 표시(Indicator) 또는 미처리** | 결측 플래그(`age_is_na`) 생성 후 별도 관리하거나 분석 목적별 선택적 제거 | • 원본 데이터의 불확실성과 무응답 패턴 자체를 보존 | • 수치형 입력만 받는 일부 알고리즘 이용 시 전처리 추가 필요 |

### 2. 처리 전후 검증 항목 (Validation Checklist)

#### 처리 전 검증 (Pre-Validation)

1. **결측 패턴(Missing Mechanism) 확인**:
* 3건의 결측 고객이 특정 도시(`city` 결측 2건과의 중복 여부), 특정 가입 경로, 특정 연령층에 치우쳐 있는지 확인 (MCAR/MAR 여부 파악).


2. **기초 통계량 및 이상치 확인**:
* 결측을 제외한 기존 `age` 수치의 최소/최대값, 음수 존재 여부, 극단치(Outlier) 유무 점검.


3. **타 변수와의 연관성 파악**:
* 결측 고객 3명의 다른 컬럼(구매액, 가입일 등) 데이터 존재 여부 확인.



#### 처리 후 검증 (Post-Validation)

1. **분포 변화 검증**:
* 대체 전/후 `age`의 **평균, 표준편차, 왜도(Skewness), 사분위수(IQR)** 비교. (대체 후 분포가 과도하게 뾰족해지거나 왜곡되지 않았는지 확인)


2. **도메인 유효성 검증**:
* 대체된 연령값이 비즈니스적으로 타당한 범위 내에 존재하는지 확인 (예: 소수점 발생 여부, 서비스 가입 가능 연령 범위 충족 여부 등).


3. **세그먼트 비중 변화 검증**:
* 10대, 20대, 30대 등 연령대 구간(Binning) 분석 시 특정 구간으로 비중이 쏠리는 현상 유무 확인.
