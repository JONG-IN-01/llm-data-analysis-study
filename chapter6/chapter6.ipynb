| 카테고리별 completed 주문 수량과 금액은 어떻게 다른가? | 매출 | completed 주문건수, 카테고리별 주문수량, 금액 | completed 만 필터링하고 데이터간 merge가 필요하다.  |

## 3. 관찰 → 가설 → 추가 검증
### 사례 1
- 관찰: 2025년 9월, 2026년 9월 판매액과 order_count가 작다. 
- 가설: 9월 판매액이 누락되었을 것이다. 또는 9월만 매출이 적은 이유가 있을 것이다.
- 추가 검증: 9월이 작은 이유가 데이터가 누락이 되었는지(ex. 1년 단위로 끊어서 1일부터 31일까지 안 들어간건지) 확인해야한다.

### 사례 2
- 관찰: 고객별 판매 금액 분석에서 수도권(경기 또는 서울)에 거주하는 사람이 다수다.
- 가설: 경기 또는 서울 사람 비중이 전체적으로 많을 것이다.
- 추가 검증: city 칼럼을 권역별로 분석해본다.

### 사례 3
- 관찰: 전체 구매금액이 높은 사람이 항상 많이 반복해서 구입한 사람은 아니다.
- 가설: 전체 구매금액이 높은 사람은 특정 카테고리(평균 판매가가 높은)에서 구입했을 것이다.
- 추가 검증: 전체 구매금액이 높은 사람이 어떤 order item을 구입했는지 하나씩 살펴본다.

## 4. LLM 질문 확장
- Safe Context: safe 하다.
- Prompt: 
온라인 쇼핑몰 데이터로 EDA를 수행하려고 합니다.
EDA는 관찰, 가정, 분석으로 구분됩니다.

데이터셋:
- customers: customer_id, gender, age, city, signup_date
- products: product_id, product_name, category, price
- orders: order_id, customer_id, order_date, payment_method, order_status
- order_items: order_id, product_id, quantity, unit_price, line_total

요청:
1. 현재 데이터로 검증 가능한 EDA 질문 10개를 만들어 주세요.
2. 각 질문에 필요한 데이터셋과 pandas 기능을 함께 정리해 주세요.
3. 데이터에 없는 내용은 추측하지 마세요.
4. 각 질문을 계산 가능한 지표로 바꿔 주세요.

- LLM 제안 요약:
질문 10개 요약
1) 월별 매출은 일정한가?
2) 매출이 소수 카테고리에 집중되어 있는가?
3) 많이 팔린 상품과 매출이 큰 상품은 같은가?
4) 결제수단에 따라 주문금액이 다른가?
5) 주문 상태 비율은 카테고리마다 다른가?
6) 연령대에 따라 주문금액이 다른가?
7) 성별에 따라 많이 사는 카테고리가 다른가?
8) 재구매 고객은 매출에 얼마나 기여하는가?
9) 가입 후 얼마나 빨리 첫 주문을 하는가?
10) 금액 데이터는 서로 맞는가? (데이터 품질 점검)

- 실제 사용 여부: 수정 후 사용
- 판단 이유: 데이터를 분석하는 사람이 생각하지 못한 내용을 제공해주는 경우가 있다. 경험이 많지 않은 사람에게는 매우 큰 도움이 될 것이다. 경험이 많은 사람 입장에서는 습과적으로 보는 수치말고도 다른 수치에 집중하게 해주는 아이디어도 제공할 것이다.
- 사람이 수정한 내용: 가입 후 얼마나 빨리 첫 주문을 하는가에 더 나아가, 2번째 주문은 언제인지까지 분석을 하면 좋을 것 같다. 주문 상태 비율은 카테고리마다 분석하는 것은 가장 중요하지 않다고 판단된다.

## 5. 다음 분석 우선순위
1. 2025년 9월 데이터와 2026년 9월 데이터의 date 날짜들을 value_count와 같은 방법을 이용하여 1일부터 30일까지 모두 포함되었는지 확인한다.
2. city 칼럼을 권역별로 분석해본다.
3. 전체 구매금액이 높은 사람이 어떤 order item을 구입했는지 하나씩 살펴본다.

### 가장 먼저 확인하고 싶은 이유
1. 9월 데이터가 1~30일까지 모두 포함되어있는데, 9월이 total 금액이 작다면 9월에 total 금액이 작은 이유를 분석할 필요가 있기 때문이다.
2. 권역별로 분석해보고 특정 권역에 집중되어있다면 이유를 생각해보기위해서이다.
3. 구매금액이 높은 고객들은 어떤 order item을 구입하였는지 살펴보고 해당 item을 적극적으로 광고할 필요가 있는지 의사결정을 하기 위해서이다.

### 현재 데이터만으로 단정할 수 없는 것
1. 9월에 total 판매 금액이 작다는 것을 단정할 수 없다.
2. 권역별로 판매금액이 다른 이유는 인구수, 접근성 등 다양한 이유가 존재하기에 단정할 수 없다.
3. order item 들의 수익성을 알 수 없다.(ex. 공헌이익률, 영업이익률)
